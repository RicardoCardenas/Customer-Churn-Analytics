```python
import pandas as pd
import numpy as np

# Ruta al archivo descargado
file_path = '../data/raw/Customer_Churn_Dataset.xlsx'

# 1. Verificar los nombres de las pestañas/hojas existentes
excel_file = pd.ExcelFile(file_path)
print("Hojas encontradas en el archivo:", excel_file.sheet_names)

# 2. Cargar los datos (por defecto lee la primera hoja, o especifica sheet_name)
df_raw = pd.read_excel(file_path, sheet_name=0)

# 3. Vista preliminar de dimensiones y columnas
print(f"Dimensiones del dataset: {df_raw.shape[0]} filas, {df_raw.shape[1]} columnas")
print("\nPrimeras 5 filas:")
display(df_raw.head())
```

```python
# Guardar copia local en CSV dentro de data/raw para lecturas rápidas posteriores
df_raw.to_csv('../data/raw/telco_churn_raw.csv', index=False)
```
